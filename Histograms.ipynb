```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
scores = np.random.normal(loc = 80, scale = 10, size = 100) #loc = centre, Scale = SD, size = no. we want to generate
scores = np.clip(scores, 0, 100) #setting min and max bounds
plt.hist(scores, 
         bins = 10,
        color = "purple",
        edgecolor = "white")

plt.title ("Exam Scores")
plt.xlabel("Scores")
plt.ylabel("Number of Students")
plt.show()
```
