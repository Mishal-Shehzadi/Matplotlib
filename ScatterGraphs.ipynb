```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 9]) #gours studied
y = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) #grades
```

```python
plt.scatter(x , y)
plt.show()
```

```python
plt.scatter(x , y)
plt.title("Test Scores")
plt.xlabel ("Hours Studied")
plt.ylabel ("Grades")
plt.show()
```

```python
plt.scatter(x , y, color = "#a83254",
           alpha = 0.5,
           s = 200)
plt.title("Test Scores")
plt.xlabel ("Hours Studied")
plt.ylabel ("Grades")
plt.show()
```

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 9]) #gours studied
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) #grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) #gours studied
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) #grades

plt.scatter(x1 , y1, color = "#a83254",
           alpha = 0.5,
           s = 50)
plt.title("Test Scores")
plt.xlabel ("Hours Studied")
plt.ylabel ("Grades")

plt.scatter(x2 , y2, color = "green",
           alpha = 0.5,
           s = 50)
plt.title("Test Scores")
plt.xlabel ("Hours Studied")
plt.ylabel ("Grades")

plt.show()
```

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 9]) #gours studied
y1 = np.array([55, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87]) #grades

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8]) #gours studied
y2 = np.array([50, 58, 65, 70, 72, 78, 83, 88, 92, 95, 97]) #grades

plt.scatter(x1 , y1, color = "#a83254",
           alpha = 0.5,
           s = 50,
           label = "Class A")

plt.scatter(x2 , y2, color = "green",
           alpha = 0.5,
           s = 50,
           label = "Class B")
plt.title("Test Scores")
plt.xlabel ("Hours Studied")
plt.ylabel ("Grades")
plt.legend()
plt.show()
```
