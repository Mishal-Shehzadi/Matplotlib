```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
df = pd.read_csv ("data.csv")
```

```python
type_count = df["Type1"].value_counts(ascending = True)
plt.barh(type_count.index, type_count.values, color = "orange",
        edgecolor = "brown")
plt.title ("Number of Pokemon by primary Type")
plt.xlabel("Count")
plt.ylabel("Type")
plt.tight_layout()
plt.show()
```
