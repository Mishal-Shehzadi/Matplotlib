```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array([4, 3, 2,  5, 3, 1])
```

```python
plt.bar(categories, values)
plt.show()
```

```python
plt.bar(categories, values)
plt.title("Daily Consumption")
plt.show()
```

```python
plt.bar(categories, values)
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

```python
plt.bar(categories, values, color = "#a83254")
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

```python
plt.barh(categories, values, color = "#a83254")
plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```
