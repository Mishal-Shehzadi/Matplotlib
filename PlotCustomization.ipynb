```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])
```

```python
plt.plot (x, y, marker = ".")
plt.show()
```

```python
plt.plot (x, y, marker = "o")
plt.show()
```

```python
plt.plot (x, y, marker = "*")
plt.show()
```

```python
plt.plot (x, y, marker = "*",
         markersize = 10)
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         ms = 10)
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "red")
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         mfc = "cyan")
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "cyan",
         markeredgecolor = "purple")
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "dashed")
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "dotted")
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "None")
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "solid",
         linewidth = 5)
plt.show()
```

```python
plt.plot (x, y, marker = "o",
         markersize = 20,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "solid",
         linewidth = 5,
         color = "blue")
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
```

```python
plt.plot (x, y1, marker = "o",
         markersize = 10,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "solid",
         linewidth = 2,
         color = "blue")
plt.plot (x, y2, marker = "o",
         markersize = 10,
         markerfacecolor = "cyan",
         markeredgecolor = "purple",
         linestyle = "solid",
         linewidth = 2,
         color = "blue")
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
```

```python
line_style = dict(marker = "o",
             markersize = 10,
             markerfacecolor = "cyan",
             markeredgecolor = "purple",
             linestyle = "solid",
             linewidth = 2,
             color = "blue")
plt.plot (x, y1, **line_style)
plt.plot (x, y2, **line_style)
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

line_style = dict(marker = "o",
             markersize = 10,
             markerfacecolor = "cyan",
             markeredgecolor = "purple",
             linestyle = "solid",
             linewidth = 2,
             color = "blue")

plt.plot (x, y1, **line_style)
plt.plot (x, y2, **line_style)
plt.plot (x, y3, **line_style)
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

line_style = dict(marker = "o",
             markersize = 10,
             markerfacecolor = "cyan",
             markeredgecolor = "purple",
             linestyle = "solid",
             linewidth = 2)

plt.plot (x, y1, color="blue", **line_style)
plt.plot (x, y2, color="#a83254", **line_style)
plt.plot (x, y3, color="green", **line_style)
```
