```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

plt.title("Class Size")

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
```

```python
plt.title("Class Size", 
          fontsize = 20,
          family = "Tahoma")
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
```

```python
plt.title("Class Size", 
          fontsize = 20,
          family = "Tahoma",
          fontweight = "bold")
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
```

```python
plt.title("Class Size", 
          fontsize = 20,
          family = "Tahoma",
          fontweight = "bold",
         color = "#a83254")
plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
```

```python
plt.title("Class Size", 
          fontsize = 20,
          family = "Tahoma",
          fontweight = "bold",
         color = "#a83254")
plt.xlabel("Year", 
           fontsize = 10,
           family = "Tahoma",
           fontweight = "bold",
           color = "#a83254")
plt.ylabel("Strength of students",
           fontsize = 10,
           family = "Tahoma",
           fontweight = "bold",
           color = "#a83254")

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)
plt.show()
```

```python
plt.title("Class Size", 
          fontsize = 20,
          family = "Tahoma",
          fontweight = "bold",
         color = "#a83254")
plt.xlabel("Year", 
           fontsize = 10,
           family = "Tahoma",
           fontweight = "bold",
           color = "#a83254")
plt.ylabel("Strength of students",
           fontsize = 10,
           family = "Tahoma",
           fontweight = "bold",
           color = "#a83254")

plt.tick_params(axis = "both",
               colors = "#a83254")

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x)
plt.show()
```
