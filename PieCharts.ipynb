```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = np.array([300, 250, 275, 225])
```

```python
plt.pie(values)
plt.show()
```

```python
plt.pie(values, labels=categories)
plt.show()
```

```python
plt.pie(values, 
        labels = categories,
       autopct = "%1.1f")
plt.show()
```

```python
plt.pie(values, 
        labels = categories,
       autopct = "%1.1f%%")
plt.show()
```

```python
colours = ["pink", "lightgreen", "yellow", "lightblue"]
plt.pie(values, 
        labels = categories,
       autopct = "%1.1f%%",
       colors = colours)
plt.show()
```

```python
colours = ["pink", "lightgreen", "yellow", "lightblue"]
plt.pie(values, 
        labels = categories,
       autopct = "%1.1f%%",
       colors = colours)
plt.title ("My College",
          color = "#a83254")
plt.show()
```

```python
colours = ["pink", "lightgreen", "yellow", "lightblue"]
plt.pie(values, 
        labels = categories,
       autopct = "%1.1f%%",
       colors = colours,
       explode = [0, 0, 0, 0.1],
       shadow = True)
plt.title ("My College")
plt.show()
```

```python
colours = ["pink", "lightgreen", "yellow", "lightblue"]
plt.pie(values, 
        labels = categories,
       autopct = "%1.1f%%",
       colors = colours,
       explode = [0, 0, 0, 0.1],
       shadow = True,
       startangle = 180)
plt.title ("My College")
plt.show()
```
