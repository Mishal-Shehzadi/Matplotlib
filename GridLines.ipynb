```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 10, 15, 20, 25]
```

```python
plt.grid()
plt.plot(x, y)
plt.show()
```

```python
plt.grid(axis="x")
plt.plot(x, y)
plt.show()
```

```python
plt.grid(axis="y")
plt.plot(x, y)
plt.show()
```

```python
plt.grid(axis = "y",
        linewidth = 2)
plt.plot(x, y)
plt.show()
```

```python
plt.grid(axis = "y",
        linewidth = 2,
        color = "lightgrey")
plt.plot(x, y)
plt.show()
```

```python
plt.grid(axis = "y",
        linewidth = 2,
        color = "lightgrey",
        linestyle = "dashed")
plt.plot(x, y)
plt.show()
```
